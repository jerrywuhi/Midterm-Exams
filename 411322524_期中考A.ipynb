## 基本需求（90 分）

**B1 規格書（20 分）：** 完成筆記本中的規格書。

**B2 系統功能（40 分）：**
- 兩張資料表：活動、報名紀錄；每次執行都重建並放入初始資料。
- AI 從一句話中抽出「動作」與「活動」。動作有：報名、取消、查詢。無法判斷時，回覆目前有哪些活動。
- 報名：同一位學生不可重複報名同一個活動；名額不足時拒絕。
- 取消：把紀錄狀態改為「已取消」並釋出名額；查不到紀錄時告知使用者。
- 查詢：回覆該活動的剩餘名額。

**B3 測試（20 分）：** AI 解讀測試 8～10 句（正常、錯誤、邊界都要有），並印出通過幾句；流程測試 8～10 步，最後顯示資料表。

**B4 交件（10 分）：** 依「交件方式」完成。

## 進階需求（30 分）

**A1 人數欄位（10 分）：** AI 再抽出第 3 個欄位「人數」（含攜伴，例如「我和兩個朋友」＝3 人），沒說時視為 1 人。人數不在 1～3 之間時由程式拒絕。名額改以**人數加總**計算。

**A2 每人上限（10 分）：** 每位學生最多同時報名 **2 個**活動（已取消的不計）。

**A3 統計查詢（10 分）：** 使用者說「各活動報名狀況」之類的話時，用 **JOIN 與 GROUP BY** 列出每個活動的日期、名額、已報名人數、剩餘名額。**沒有人報名的活動也必須列出。**

````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('jerry')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 功能：
# 1. 建立 SQLite 資料庫
# 2. 刪除舊資料表並重新建立
# 3. 建立活動資料表與報名紀錄資料表
# 4. 新增三筆初始活動資料
# ============================================================

# 建立資料庫連線
conn = sqlite3.connect("activity.db")

try:
    cursor = conn.cursor()

    # 啟用外鍵功能
    cursor.execute("PRAGMA foreign_keys = ON")

    # 刪除舊資料表
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS activities")

    # 建立活動資料表
    cursor.execute("""
        CREATE TABLE activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL UNIQUE,
            date TEXT NOT NULL,
            capacity INTEGER NOT NULL
        )
    """)

    # 建立報名紀錄資料表
    cursor.execute("""
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            activity_id INTEGER NOT NULL,
            people INTEGER NOT NULL,
            status TEXT NOT NULL DEFAULT '已報名',
            FOREIGN KEY (activity_id) REFERENCES activities(id)
        )
    """)

    # 新增三場初始活動
    initial_activities = [
        ("迎新晚會", "2026-11-20", 6),
        ("程式工作坊", "2026-11-25", 4),
        ("籃球友誼賽", "2026-11-28", 3)
    ]

    cursor.executemany("""
        INSERT INTO activities (name, date, capacity)
        VALUES (?, ?, ?)
    """, initial_activities)

    # 儲存資料
    conn.commit()

    print("✅ 資料庫建立成功！")
    print("✅ 已建立 activities 活動資料表")
    print("✅ 已建立 registrations 報名紀錄資料表")
    print("✅ 已新增三場初始活動")

except Exception as e:
    conn.rollback()
    print("❌ 資料庫建立失敗：", e)

finally:
    conn.close()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 功能：
# 1. 使用 Groq AI 辨識使用者需求
# 2. 判斷動作、活動及報名人數
# 3. 使用 Python 驗證 AI 回覆
# 4. 加入 API 錯誤處理及 429 重試機制
# 5. 使用字典快取，避免重複呼叫 AI
# ============================================================

# 建立 AI 解讀結果快取
ai_cache = {}

# 設定允許的動作及活動
ALLOWED_ACTIONS = ["報名", "取消", "查詢", "統計"]

ALLOWED_ACTIVITIES = [
    "迎新晚會",
    "程式工作坊",
    "籃球友誼賽",
    "全部"
]


def ai_interpret(user_input):

    # 去除前後空白
    user_input = user_input.strip()

    # 空白輸入直接回傳未知
    if not user_input:
        return "未知", "未知", 1, "使用者沒有輸入內容"

    # 檢查快取，避免相同句子重複呼叫 AI
    if user_input in ai_cache:
        return ai_cache[user_input]

    # 建立 AI 提示詞
    system_prompt = """
你是一個系學會活動報名系統的 AI 助手。

你的工作是分析使用者輸入的句子，
並辨識以下三個欄位：

1. 動作
只能是：報名、取消、查詢、統計。
如果無法判斷，輸出未知。

2. 活動
只能是：迎新晚會、程式工作坊、籃球友誼賽、全部。
如果無法判斷，輸出未知。
查詢所有活動報名狀況時，活動輸出全部。

3. 人數
沒有提到人數時預設為 1。
我和一個朋友代表 2 人。
我和兩個朋友代表 3 人。
我和三個朋友代表 4 人。
請輸出阿拉伯數字。

請注意：
- 只輸出三個欄位。
- 不要輸出解釋。
- 不要使用 Markdown。
- 不要輸出其他文字。

輸出格式：
動作,活動,人數

例如：
報名,迎新晚會,1
取消,程式工作坊,1
查詢,籃球友誼賽,1
報名,迎新晚會,3
統計,全部,1
未知,未知,1
"""

    raw_reply = ""

    # 最多呼叫 AI 三次
    for attempt in range(3):

        try:
            # 呼叫 Groq AI
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "system",
                        "content": system_prompt
                    },
                    {
                        "role": "user",
                        "content": user_input
                    }
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            # 取得 AI 原始回覆
            raw_reply = response.choices[0].message.content or ""

            # 每次呼叫後等待 2 秒
            time.sleep(2)

            # 成功取得結果後離開重試迴圈
            break

        except Exception as e:

            # 每次呼叫失敗後也等待 2 秒
            time.sleep(2)

            error_message = str(e)

            # 判斷是否為 API 使用量超過限制
            if (
                "429" in error_message
                or "rate_limit" in error_message.lower()
                or "too many requests" in error_message.lower()
            ):

                # 還有重試次數時等待 30 秒
                if attempt < 2:
                    print(f"⚠️ API 使用量超過限制，等待 30 秒後重試（第 {attempt + 2} 次）")
                    time.sleep(30)
                    continue

                raw_reply = "AI 呼叫失敗：已達最大重試次數"

            else:
                raw_reply = "AI 呼叫失敗：" + error_message
                break

    # 如果 AI 呼叫失敗，回傳未知
    if raw_reply.startswith("AI 呼叫失敗"):
        result = ("未知", "未知", 1, raw_reply)
        ai_cache[user_input] = result
        return result

    # --------------------------------------------------------
    # 以下使用 Python 驗證 AI 回覆
    # 不直接相信 AI 提供的內容
    # --------------------------------------------------------

    # 移除多餘空白及 Markdown 標記
    clean_reply = raw_reply.strip()
    clean_reply = clean_reply.replace("```", "").strip()

    # 依照逗號切分欄位
    parts = re.split(r"[,，|]+", clean_reply)

    # 預設全部為未知
    action = "未知"
    activity = "未知"
    people = 1

    # 辨識動作
    # 優先檢查第一個欄位，避免其他文字造成誤判
    action_text = parts[0] if len(parts) >= 1 else clean_reply

    for item in ALLOWED_ACTIONS:
        if item in action_text:
            action = item
            break

    # 如果第一個欄位沒有找到，再從完整回覆尋找
    if action == "未知":
        for item in ALLOWED_ACTIONS:
            if item in clean_reply:
                action = item
                break

    # 辨識活動
    activity_text = parts[1] if len(parts) >= 2 else clean_reply

    for item in ALLOWED_ACTIVITIES:
        if item in activity_text:
            activity = item
            break

    # 如果第二個欄位沒有找到，再從完整回覆尋找
    if activity == "未知":
        for item in ALLOWED_ACTIVITIES:
            if item in clean_reply:
                activity = item
                break

    # 使用正規表示式尋找人數
    # 優先從第三個欄位取得數字
    people_text = parts[2] if len(parts) >= 3 else clean_reply

    numbers = re.findall(r"\d+", people_text)

    # 找不到數字時，預設為 1
    if numbers:
        people = int(numbers[0])
    else:
        people = 1

    # 組合最後的解讀結果
    result = (action, activity, people, raw_reply)

    # 儲存到快取
    ai_cache[user_input] = result

    return result


print("✅ AI 解讀函式建立完成！")
print("✅ 已設定動作及活動允許清單")
print("✅ 已加入人數辨識功能")
print("✅ 已加入 API 重試及快取機制")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 功能：
# 1. 使用 9 句測試資料測試 AI 解讀
# 2. 顯示 AI 原始回覆
# 3. 顯示系統解讀結果
# 4. 比對預期答案並統計通過數量
# ============================================================

# 測試資料：(使用者說的話, 預期動作, 預期活動, 預期人數)
ai_test_cases = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),

    ("幫我取消程式工作坊", "取消", "程式工作坊", 1),

    ("籃球友誼賽還剩多少名額？", "查詢", "籃球友誼賽", 1),

    ("我和兩個朋友要報名迎新晚會", "報名", "迎新晚會", 3),

    ("查詢各活動報名狀況", "統計", "全部", 1),

    ("我要報名游泳比賽", "未知", "未知", 1),

    ("今天天氣很好", "未知", "未知", 1),

    ("我和一個朋友報名程式工作坊", "報名", "程式工作坊", 2),

    ("我和三個朋友報名籃球友誼賽", "報名", "籃球友誼賽", 4)
]

# 初始化通過數量
passed = 0
total = len(ai_test_cases)

print("========== AI 解讀測試開始 ==========")

# 逐一執行測試
for i, test_case in enumerate(ai_test_cases, start=1):

    user_input, expected_action, expected_activity, expected_people = test_case

    print(f"\n========== 第 {i} 題 ==========")
    print("使用者輸入：", user_input)

    # 呼叫 AI 解讀函式
    action, activity, people, raw_reply = ai_interpret(user_input)

    # 顯示 AI 原始回覆
    print("AI 原始回覆：", raw_reply)

    # 顯示系統解讀
    print("系統解讀：")
    print("動作：", action)
    print("活動：", activity)
    print("人數：", people)

    # 比對預期答案
    if (
        action == expected_action
        and activity == expected_activity
        and people == expected_people
    ):
        print("測試結果：✅ 通過")
        passed += 1
    else:
        print("測試結果：❌ 未通過")
        print("預期結果：",
              expected_action,
              expected_activity,
              expected_people)

# 顯示測試統計
print("\n========== AI 解讀測試結果 ==========")
print("總測試數量：", total)
print("通過數量：", passed)
print("失敗數量：", total - passed)
print("通過率：", round(passed / total * 100, 2), "%")

if passed == total:
    print("🎉 所有 AI 解讀測試皆通過！")
else:
    print("⚠️ 部分測試未通過，請檢查 AI 解讀結果。")
````

````python
# ============================================================
# 第 6 格：功能函式
# 功能：
# 1. 活動報名
# 2. 取消報名
# 3. 查詢剩餘名額
# 4. 統計所有活動報名狀況
# 5. 限制每人最多報名兩個活動
# 6. 限制每次報名人數為 1～3 人
# ============================================================

def process_request(action, activity, people, student_name=""):

    # 清除姓名前後空白
    student_name = student_name.strip()

    # 建立資料庫連線
    conn = sqlite3.connect("activity.db")

    try:
        cursor = conn.cursor()

        # ====================================================
        # 1. 無法判斷時，顯示目前所有活動
        # ====================================================
        if action == "未知" or activity == "未知":

            cursor.execute("""
                SELECT name, date, capacity
                FROM activities
            """)

            activities = cursor.fetchall()

            result = "無法判斷您的需求，目前提供以下活動：\n"

            for item in activities:
                result += (
                    f"\n活動：{item[0]}"
                    f"\n日期：{item[1]}"
                    f"\n名額：{item[2]} 人\n"
                )

            return result

        # ====================================================
        # 2. 統計所有活動
        # 使用 LEFT JOIN 與 GROUP BY
        # ====================================================
        if action == "統計":

            statistics = pd.read_sql_query("""
                SELECT
                    a.name AS 活動名稱,
                    a.date AS 活動日期,
                    a.capacity AS 活動名額,
                    COALESCE(
                        SUM(
                            CASE
                                WHEN r.status = ?
                                THEN r.people
                                ELSE 0
                            END
                        ), 0
                    ) AS 已報名人數,
                    a.capacity - COALESCE(
                        SUM(
                            CASE
                                WHEN r.status = ?
                                THEN r.people
                                ELSE 0
                            END
                        ), 0
                    ) AS 剩餘名額
                FROM activities a
                LEFT JOIN registrations r
                    ON a.id = r.activity_id
                GROUP BY
                    a.id,
                    a.name,
                    a.date,
                    a.capacity
                ORDER BY a.id
            """, conn, params=("已報名", "已報名"))

            return "===== 各活動報名統計 =====\n\n" + statistics.to_string(index=False)

        # ====================================================
        # 3. 檢查活動是否存在
        # ====================================================
        if activity not in ALLOWED_ACTIVITIES or activity == "全部":
            return "❌ 活動名稱錯誤，請選擇目前提供的三場活動。"

        cursor.execute("""
            SELECT id, name, date, capacity
            FROM activities
            WHERE name = ?
        """, (activity,))

        activity_data = cursor.fetchone()

        # fetchone() 後先檢查是否為 None
        if activity_data is None:
            return "❌ 查無此活動。"

        activity_id = activity_data[0]
        capacity = activity_data[3]

        # ====================================================
        # 4. 查詢剩餘名額
        # ====================================================
        if action == "查詢":

            cursor.execute("""
                SELECT COALESCE(SUM(people), 0)
                FROM registrations
                WHERE activity_id = ?
                AND status = ?
            """, (activity_id, "已報名"))

            registered_data = cursor.fetchone()

            if registered_data is None:
                registered_people = 0
            else:
                registered_people = registered_data[0]

            remaining = capacity - registered_people

            return (
                f"活動：{activity}\n"
                f"活動日期：{activity_data[2]}\n"
                f"總名額：{capacity} 人\n"
                f"已報名：{registered_people} 人\n"
                f"剩餘名額：{remaining} 人"
            )

        # ====================================================
        # 5. 報名與取消需要姓名
        # ====================================================
        if action in ["報名", "取消"] and not student_name:
            return "❌ 請輸入學生姓名，才能進行報名或取消。"

        # ====================================================
        # 6. 取消報名
        # ====================================================
        if action == "取消":

            cursor.execute("""
                SELECT id
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
                AND status = ?
                ORDER BY id DESC
                LIMIT 1
            """, (student_name, activity_id, "已報名"))

            registration = cursor.fetchone()

            # 檢查是否找到報名紀錄
            if registration is None:
                return "❌ 查無報名紀錄，無法取消。"

            # 將報名狀態修改為已取消
            cursor.execute("""
                UPDATE registrations
                SET status = ?
                WHERE id = ?
            """, ("已取消", registration[0]))

            conn.commit()

            return (
                f"✅ 取消成功！\n"
                f"學生：{student_name}\n"
                f"活動：{activity}\n"
                f"已釋出 {people if people else 1} 個名額。"
            )

        # ====================================================
        # 7. 報名功能
        # ====================================================
        if action == "報名":

            # 檢查報名人數是否介於 1～3 人
            if not isinstance(people, int) or not 1 <= people <= 3:
                return "❌ 報名人數必須介於 1～3 人。"

            # 開始資料庫交易，避免同時報名造成超額
            cursor.execute("BEGIN IMMEDIATE")

            # 檢查是否已經報名同一活動
            cursor.execute("""
                SELECT id
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
                AND status = ?
            """, (student_name, activity_id, "已報名"))

            existing = cursor.fetchone()

            if existing is not None:
                conn.rollback()
                return "❌ 您已經報名此活動，不可重複報名。"

            # 檢查目前已報名的不同活動數量
            cursor.execute("""
                SELECT COUNT(DISTINCT activity_id)
                FROM registrations
                WHERE student_name = ?
                AND status = ?
            """, (student_name, "已報名"))

            count_data = cursor.fetchone()

            if count_data is None:
                current_count = 0
            else:
                current_count = count_data[0]

            # 每位學生最多同時報名兩個活動
            if current_count >= 2:
                conn.rollback()
                return "❌ 您已達最多同時報名兩個活動的上限。"

            # 計算目前活動已報名人數
            cursor.execute("""
                SELECT COALESCE(SUM(people), 0)
                FROM registrations
                WHERE activity_id = ?
                AND status = ?
            """, (activity_id, "已報名"))

            registered_data = cursor.fetchone()

            if registered_data is None:
                registered_people = 0
            else:
                registered_people = registered_data[0]

            # 檢查剩餘名額
            remaining = capacity - registered_people

            if people > remaining:
                conn.rollback()
                return (
                    f"❌ 報名失敗！\n"
                    f"活動剩餘名額只有 {remaining} 人，"
                    f"無法報名 {people} 人。"
                )

            # 新增報名紀錄
            cursor.execute("""
                INSERT INTO registrations
                (student_name, activity_id, people, status)
                VALUES (?, ?, ?, ?)
            """, (student_name, activity_id, people, "已報名"))

            # 儲存資料
            conn.commit()

            return (
                f"✅ 報名成功！\n"
                f"學生：{student_name}\n"
                f"活動：{activity}\n"
                f"報名人數：{people} 人\n"
                f"報名後剩餘名額：{remaining - people} 人"
            )

        # ====================================================
        # 8. 無法辨識的動作
        # ====================================================
        return "❌ 無法辨識操作，請重新輸入需求。"

    except Exception as e:

        # 發生錯誤時取消尚未完成的交易
        conn.rollback()

        return "❌ 系統發生錯誤：" + str(e)

    finally:

        # 確保資料庫連線關閉
        conn.close()


print("✅ 功能函式建立完成！")
print("✅ 已完成報名、取消及查詢功能")
print("✅ 已加入人數限制及每人兩場活動限制")
print("✅ 已完成 JOIN + GROUP BY 統計功能")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 功能：
# 1. 整合 AI 解讀與系統功能
# 2. 建立主流程函式
# 3. 執行 10 步連續流程測試
# 4. 顯示每一步執行結果
# 5. 最後顯示資料表
# ============================================================


# ------------------------------------------------------------
# 一、建立系統主流程
# ------------------------------------------------------------

def run_system(user_input, student_name=""):

    # 使用 AI 解讀使用者需求
    action, activity, people, raw_reply = ai_interpret(user_input)

    print("AI 原始回覆：", raw_reply)
    print("系統解讀：", action, activity, people)

    # 將解讀結果交給功能函式執行
    result = process_request(
        action,
        activity,
        people,
        student_name
    )

    return result


# ------------------------------------------------------------
# 二、設定流程測試資料
# 格式：(姓名, 使用者說的話, 預期結果)
# ------------------------------------------------------------

flow_test_cases = [

    (
        "學生A",
        "我要報名迎新晚會",
        "報名成功"
    ),

    (
        "學生A",
        "我要報名程式工作坊",
        "報名成功"
    ),

    (
        "學生A",
        "我要報名籃球友誼賽",
        "已達最多同時報名兩個活動的上限"
    ),

    (
        "學生A",
        "幫我取消迎新晚會",
        "取消成功"
    ),

    (
        "學生A",
        "我要報名籃球友誼賽",
        "報名成功"
    ),

    (
        "學生A",
        "我要再次報名籃球友誼賽",
        "不可重複報名"
    ),

    (
        "學生A",
        "幫我取消程式工作坊",
        "取消成功"
    ),

    (
        "學生A",
        "我要和兩個朋友報名迎新晚會",
        "報名成功"
    ),

    (
        "",
        "迎新晚會還剩多少名額",
        "剩餘名額：3 人"
    ),

    (
        "",
        "各活動報名狀況",
        "各活動報名統計"
    )

]


# ------------------------------------------------------------
# 三、執行流程測試
# ------------------------------------------------------------

passed = 0
total = len(flow_test_cases)

print("========== 系學會活動報名系統流程測試 ==========")

for i, test_case in enumerate(flow_test_cases, start=1):

    student_name, user_input, expected = test_case

    print(f"\n========== 第 {i} 步 ==========")
    print("學生姓名：", student_name if student_name else "不需要")
    print("使用者輸入：", user_input)

    # 執行系統主流程
    result = run_system(user_input, student_name)

    print("\n系統回應：")
    print(result)

    # 檢查系統回應是否符合預期
    if expected in result:
        print("測試結果：✅ 通過")
        passed += 1

    else:
        print("測試結果：❌ 未通過")
        print("預期結果：", expected)


# ------------------------------------------------------------
# 四、顯示流程測試統計
# ------------------------------------------------------------

print("\n========== 流程測試統計 ==========")

print("總測試步數：", total)
print("通過步數：", passed)
print("失敗步數：", total - passed)

print("通過率：", round(passed / total * 100, 2), "%")

if passed == total:
    print("🎉 所有流程測試皆通過！")
else:
    print("⚠️ 部分流程測試未通過，請檢查程式。")


# ------------------------------------------------------------
# 五、顯示最後資料表
# ------------------------------------------------------------

print("\n========== 最終資料庫內容 ==========")

show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 功能：
# 1. 建立活動報名系統網頁
# 2. 輸入使用者需求及姓名
# 3. 呼叫 AI 解讀與系統功能
# 4. 顯示系統執行結果
# ============================================================


# ------------------------------------------------------------
# 一、建立網頁主要處理函式
# ------------------------------------------------------------

def gradio_process(user_input, student_name):

    # 檢查使用者是否輸入需求
    if not user_input or not user_input.strip():
        return "❌ 請輸入您的需求。"

    # 將空白姓名轉換為空字串
    if student_name is None:
        student_name = ""

    # 執行系統主流程
    result = run_system(user_input, student_name)

    return result


# ------------------------------------------------------------
# 二、建立 Gradio 網頁介面
# ------------------------------------------------------------

with gr.Blocks(title="系學會活動報名系統") as demo:

    gr.Markdown(
        """
        # 🎓 系學會活動報名系統

        歡迎使用系學會活動報名系統！

        您可以透過自然語言完成活動報名、取消及名額查詢。

        ### 📌 目前提供的活動

        | 活動名稱 | 活動日期 | 名額 |
        |---|---|---:|
        | 迎新晚會 | 2026-11-20 | 6 人 |
        | 程式工作坊 | 2026-11-25 | 4 人 |
        | 籃球友誼賽 | 2026-11-28 | 3 人 |

        ### 💡 使用範例

        - 我要報名迎新晚會
        - 幫我取消程式工作坊
        - 我和兩個朋友要報名迎新晚會
        - 籃球友誼賽還剩多少名額？
        - 各活動報名狀況
        """
    )

    # 學生姓名輸入框
    student_name_input = gr.Textbox(
        label="學生姓名",
        placeholder="報名或取消時請輸入姓名"
    )

    # 使用者需求輸入框
    user_input = gr.Textbox(
        label="請輸入您的需求",
        placeholder="例如：我要報名迎新晚會",
        lines=2
    )

    # 建立按鈕
    with gr.Row():

        submit_button = gr.Button(
            "送出需求",
            variant="primary"
        )

        clear_button = gr.ClearButton(
            components=[
                student_name_input,
                user_input
            ],
            value="清除"
        )

    # 系統回應區域
    output = gr.Textbox(
        label="系統回應",
        lines=10,
        interactive=False
    )

    # 按下送出按鈕後執行
    submit_button.click(
        fn=gradio_process,
        inputs=[
            user_input,
            student_name_input
        ],
        outputs=output
    )

    gr.Markdown(
        """
        ---
        **系統說明**

        - 報名及取消需要輸入姓名。
        - 每位學生最多同時報名兩個活動。
        - 每次報名人數限制為 1～3 人。
        - 活動名額以實際報名人數計算。
        """
    )


# ------------------------------------------------------------
# 三、啟動 Gradio 網頁
# ------------------------------------------------------------

print("✅ 系學會活動報名系統網頁建立完成！")

if IN_COLAB:
    demo.launch(share=True)
else:
    demo.launch()
````
